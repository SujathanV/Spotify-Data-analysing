```python

df = pd.read_csv('Spotify-Dashboard.csv', index_col=0)

if 'Unnamed: 0' in df.columns:
    df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
df.head()
```

```python
print("Missing values")
print(df.isnull().sum())

df = df.dropna()
print("\n fixed missing values",df.isnull().sum().sum())
```

```python
print("Duplicates before:",df.duplicated().sum())
df = df.drop_duplicates(subset=["track_id"])
print("Duplicates after:",df.duplicated().sum())
```

```python
df["duration_min"] = (df["duration_ms"] / 60000).round(2)

df = df.drop(columns=["duration_ms"])
print("Duration converted to minutes")
df[["track_name", "duration_min"]].head()
```

```python
df["track_name"] = df["track_name"].str.strip()
df["artists"]    = df["artists"].str.strip()
df["track_genre"]= df["track_genre"].str.strip().str.lower()

print("Text cleaned!")
```

```python
def popularity_category(score):
    if score >= 75:
        return "Hot"
    elif score >= 50:
        return "Popular"
    elif score >= 25:
        return "Rising"
    else:
        return "Low"

df["popularity_category"] = df["popularity"].apply(popularity_category)
print("Popularity categories added")
df[["track_name", "popularity", "popularity_category"]].head(10)
```

```python
print("="*45)
print("CLEANING COMPLETE!")
print("="*45)
print(f"Total songs:       {len(df):,}")
print(f"Total columns:     {len(df.columns)}")
print(f"Missing values:    {df.isnull().sum().sum()}")
print(f"Duplicates:        {df.duplicated().sum()}")
print(f"\nFinal columns:")
for col in df.columns:
    print(f"  → {col}")
```

```python
df.to_csv("spotify_cleaned.csv", index=False)
print("Saved as spotify_cleaned.csv!")
```
